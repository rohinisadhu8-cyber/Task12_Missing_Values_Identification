```python

import pandas as pd

df = pd.read_csv('/content/titanic.csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python

print("Dataset information:")
df.info()

print("\nTotal rows and columns:")
print(df.shape)
```

```python

missing_count = df.isnull().sum()

print("Missing values in each column:")
print(missing_count)
```

```python

summary = pd.DataFrame({
    "Column Name": df.columns,
    "Missing Values": df.isnull().sum().values,
    "Missing Percentage": (
        df.isnull().mean().values * 100
    ).round(2)
})

summary = summary.sort_values(
    by="Missing Values",
    ascending=False
)

print(summary.to_string(index=False))
```

Findings

Conclusion

Missing values were identified and summarized for each column using Python and Pandas. Values were not deleted or replaced without justification. This analysis helps prepare the dataset for reliable data analysis.
